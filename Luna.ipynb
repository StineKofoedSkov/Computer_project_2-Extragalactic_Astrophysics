```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
files = ['A6V.asc', 'B1p5V.asc', 'B6V.asc', 'F0V.asc', 'G6V.asc', 'K4V.asc', 'M0V.asc', 'M5V.asc', 'O5V.asc', 'O8V.asc']

for spectra in files:
    data = pd.read_csv(spectra, sep=r"\s+", header=None, names=['wavelength', 'flux_density'])
    plt.plot(data['wavelength'], data['flux_density'])
    plt.xlabel('Wavelength (Angstroms)')
    plt.ylabel('Flux Density (erg/s/cm^2/Angstrom)')
    plt.title(f'Spectrum of {spectra[:-4]} Star')
    plt.show()
```
